```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import random
```

```python
old_dataset = "breast_cancer_detection/datasetji"
new_dataset = "breast_cancer_detection/datasetji_balanced"

old_benign = os.path.join(old_dataset, "Benign")
old_malignant = os.path.join(old_dataset, "Malignant")

new_benign = os.path.join(new_dataset, "Benign")
new_malignant = os.path.join(new_dataset, "Malignant")
```

```python
def count_images(folder_path):
    return len(os.listdir(folder_path))

data = {
    "Dataset": ["Original", "Original", "Balanced", "Balanced"],
    "Class": ["Benign", "Malignant", "Benign", "Malignant"],
    "Count": [
        count_images(old_benign), count_images(old_malignant),
        count_images(new_benign), count_images(new_malignant)
    ]
}
count_df = pd.DataFrame(data)

plt.figure(figsize=(8,5))
sns.barplot(data=count_df, x="Class", y="Count", hue="Dataset", palette="coolwarm")
plt.title("Class Distribution Before and After Balancing")
plt.show()

count_df
```

```python
def show_samples(folder_path, class_name, dataset_name, n=4):
    plt.figure(figsize=(12, 3))
    for i in range(n):
        img_name = random.choice(os.listdir(folder_path))
        img_path = os.path.join(folder_path, img_name)
        img = Image.open(img_path)
        plt.subplot(1, n, i+1)
        plt.imshow(img)
        plt.axis('off')
        plt.title(f"{dataset_name} - {class_name}")
    plt.show()

show_samples(old_benign, "Benign", "Original")
show_samples(new_benign, "Benign", "Balanced")
show_samples(old_malignant, "Malignant", "Original")
show_samples(new_malignant, "Malignant", "Balanced")
```

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
data_40X = "40X"
data_100X = "100X"
data_200X = "200X"
data_400X = "400X"
```

```python
classes = ['Benign', 'Malignant']
```

# 40X DATA


```python
class_counts = {}
for cls in classes:
    folder_path = os.path.join(data_40X, cls)
    files = os.listdir(folder_path)

    images = [f for f in files if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    class_counts[cls] = len(images)

print("Class count:", class_counts)
```

```python
plt.bar(class_counts.keys(), class_counts.values(), color=['blue', 'red'])
plt.xlabel('Classes')
plt.ylabel('Number of Images')
plt.title('Class Distribution in 40X Dataset')
plt.show()
```

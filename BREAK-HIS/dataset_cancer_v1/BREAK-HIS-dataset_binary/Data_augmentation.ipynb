```python
import os
from PIL import Image
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import random
```

# 40X Augmentation

```python
input_root = "40X"
output_root = "Augmented_40X"
```

```python
os.makedirs(f"{output_root}/Benign", exist_ok=True)
os.makedirs(f"{output_root}/Malignant", exist_ok=True)
```

```python
benign_transform = transforms.Compose([
    transforms.RandomRotation(30),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.ColorJitter(brightness=0.3, contrast=0.3),
    transforms.RandomResizedCrop(224, scale=(0.7, 1.0)),
    transforms.GaussianBlur(kernel_size=3),
    transforms.ToTensor()
])
```

```python
malignant_transform = transforms.Compose([
    transforms.RandomRotation(15),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.15, contrast=0.15),
    transforms.Resize((224, 224)),
    transforms.ToTensor()
])
```

```python
dataset = datasets.ImageFolder(input_root, transform=transforms.ToTensor())
class_to_idx = dataset.class_to_idx
```

```python
benign_idx = [i for i, (_, c) in enumerate(dataset.samples) if c == class_to_idx['benign']]
malignant_idx = [i for i, (_, c) in enumerate(dataset.samples) if c == class_to_idx['malignant']]
```

```python
def generate_augmented_images(indices, transform, target_count, class_name):
    img_counter = 0
    while img_counter < target_count:
        idx = random.choice(indices)
        img_path, _ = dataset.samples[idx]
        img = Image.open(img_path).convert("RGB")

        aug_img = transform(img)
        aug_img = transforms.ToPILImage()(aug_img)

        save_path = f"{output_root}/{class_name}/{class_name}_{img_counter}.png"
        aug_img.save(save_path)

        img_counter += 1

    print(f"{class_name} augmentation done — Generated: {target_count} images")
```

```python
generate_augmented_images(benign_idx, benign_transform, 1400, "Benign")
```

```python
generate_augmented_images(malignant_idx, malignant_transform, 700, "Malignant")
```
